## Import thư viện

```python
import numpy as np
import pandas as pd

# Đặt seed để kết quả ngẫu nhiên có thể lặp lại khi kiểm tra.
rng = np.random.default_rng(42)
np.set_printoptions(suppress=True)
```

## Phần A: Bài tập NumPy

### Câu 1
Tạo mảng số nguyên ngẫu nhiên kích thước 3x3 trong khoảng `[0, 20]`, sau đó kiểm tra tất cả phần tử có khác 0 hay không bằng `numpy.all`.

```python
arr1 = rng.integers(0, 21, size=(3, 3))

print("Mảng arr1:")
print(arr1)
print("Tất cả phần tử khác 0:", np.all(arr1 != 0))
```

### Câu 2
Tạo mảng số nguyên ngẫu nhiên kích thước 3x3 trong khoảng `[0, 20]`, sau đó kiểm tra có tồn tại phần tử khác 0 hay không bằng `numpy.any`.

```python
arr2 = rng.integers(0, 21, size=(3, 3))

print("Mảng arr2:")
print(arr2)
print("Có ít nhất một phần tử khác 0:", np.any(arr2 != 0))
```

### Câu 3
Minh họa các hàm so sánh `numpy.greater`, `numpy.greater_equal`, `numpy.less`, `numpy.less_equal` trên hai mảng 1 chiều và hai mảng 2 chiều.

```python
a1 = np.array([1, 4, 7, 10])
b1 = np.array([2, 4, 6, 12])

a2 = np.array([[1, 5, 9], [3, 7, 11]])
b2 = np.array([[1, 4, 10], [4, 7, 8]])

print("So sánh hai mảng 1 chiều:")
print("a1:", a1)
print("b1:", b1)
print("greater:", np.greater(a1, b1))
print("greater_equal:", np.greater_equal(a1, b1))
print("less:", np.less(a1, b1))
print("less_equal:", np.less_equal(a1, b1))

print("\nSo sánh hai mảng 2 chiều:")
print("a2:")
print(a2)
print("b2:")
print(b2)
print("greater:")
print(np.greater(a2, b2))
print("greater_equal:")
print(np.greater_equal(a2, b2))
print("less:")
print(np.less(a2, b2))
print("less_equal:")
print(np.less_equal(a2, b2))
```

### Câu 4
Tạo ba mảng gồm 10 số 1, 10 số 0 và 10 số 5, sau đó nối lại thành một mảng bằng `numpy.concatenate`.

```python
ones = np.ones(10, dtype=int)
zeros = np.zeros(10, dtype=int)
fives = np.full(10, 5, dtype=int)
combined = np.concatenate([ones, zeros, fives])

print("Mảng sau khi nối:")
print(combined)
```

### Câu 7

### Câu 8
Tạo mảng 20 phần tử trong khoảng `[0, 20]` và đổi dấu các số nằm trong khoảng `[9, 15]`.

```python
arr8 = np.arange(20)
arr8_changed = arr8.copy()
mask = (arr8_changed >= 9) & (arr8_changed <= 15)
arr8_changed[mask] *= -1

print("Mảng ban đầu:")
print(arr8)
print("Mảng sau khi đổi dấu các số từ 9 đến 15:")
print(arr8_changed)
```

## Phần B: Bài tập Pandas

### Câu 15
Dùng Pandas để cộng, trừ, nhân và chia hai Series.

```python
s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 10])

print("Series 1:")
print(s1)
print("\nSeries 2:")
print(s2)
print("\nCộng:")
print(s1 + s2)
print("\nTrừ:")
print(s1 - s2)
print("\nNhân:")
print(s1 * s2)
print("\nChia:")
print(s1 / s2)
```

### Câu 16
Chuyển cột đầu tiên của DataFrame thành Series. Lưu ý: `ix` đã bị loại bỏ, nên dùng `iloc[:, 0]`.

```python
df16 = pd.DataFrame({
    "A": [10, 20, 30, 40],
    "B": [1.5, 2.5, 3.5, 4.5],
    "C": ["x", "y", "z", "t"]
})
first_column = df16.iloc[:, 0]

print("DataFrame ban đầu:")
print(df16)
print("\nCột đầu tiên dưới dạng Series:")
print(first_column)
```

### Câu 17
Tạo DataFrame 4x3 ngẫu nhiên và sắp xếp dữ liệu theo một cột.

```python
df17 = pd.DataFrame(
    rng.integers(1, 100, size=(4, 3)),
    columns=["col1", "col2", "col3"]
)
sorted_df17 = df17.sort_values(by="col2")

print("DataFrame ban đầu:")
print(df17)
print("\nDataFrame sau khi sắp xếp theo col2:")
print(sorted_df17)
```

### Câu 18
Thay đổi index của một Series bằng phương thức `reindex` trong Pandas.

```python
sr18 = pd.Series([10, 20, 30], index=["a", "b", "c"])
reindexed_sr18 = sr18.reindex(["c", "b", "a", "d"], fill_value=0)

print("Series ban đầu:")
print(sr18)
print("\nSeries sau khi reindex:")
print(reindexed_sr18)
```

### Câu 19
Tìm các phần tử trong Series x mà không có trong Series y bằng `isin`.

```python
sr1 = pd.Series([1, 2, 3, 4, 5])
sr2 = pd.Series([2, 4, 6, 8, 10])
result19 = sr1[~sr1.isin(sr2)]

print("Series x:")
print(sr1.to_list())
print("Series y:")
print(sr2.to_list())
print("Các phần tử thuộc x nhưng không thuộc y:")
print(result19.to_list())
```

### Câu 20
Minh họa phép hội bằng `numpy.union1d` và phép giao bằng `numpy.intersect1d` trên hai Series ngẫu nhiên có 20 phần tử.

```python
x20 = pd.Series(rng.integers(0, 30, size=20))
y20 = pd.Series(rng.integers(0, 30, size=20))

union_xy = np.union1d(x20, y20)
intersection_xy = np.intersect1d(x20, y20)

print("Series x:")
print(x20.to_list())
print("Series y:")
print(y20.to_list())
print("Phép hội union1d:")
print(union_xy)
print("Phép giao intersect1d:")
print(intersection_xy)
```

### Câu 21
Thêm Series y vào Series x theo chiều dọc và chiều ngang. Lưu ý: `Series.append` đã bị loại bỏ trong Pandas mới, nên dùng `pd.concat`.

```python
x21 = pd.Series(["A", "B", "C"], name="x")
y21 = pd.Series(["D", "E", "F"], name="y")

vertical_concat = pd.concat([x21, y21], ignore_index=True)
horizontal_concat = pd.concat([x21, y21], axis=1)

print("Series x:")
print(x21)
print("\nSeries y:")
print(y21)
print("\nNối theo chiều dọc:")
print(vertical_concat)
print("\nNối theo chiều ngang:")
print(horizontal_concat)
```

### Câu 22

```python
exam_data = {
    "name": ["Anastasia", "Dima", "Katherine", "James", "Emily", "Michael", "Matthew", "Laura", "Kevin", "Jonas"],
    "score": [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    "attempts": [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "qualify": ["yes", "no", "yes", "no", "no", "yes", "yes", "no", "no", "yes"]
}
labels = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]
df_exam = pd.DataFrame(exam_data, index=labels)

print("DataFrame exam_data:")
print(df_exam)
```
